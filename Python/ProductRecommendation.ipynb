```python
import pandas as pd
import numpy as np

# Data is imported from the real database
product_dataset = pd.read_csv('/content/Product_Dataset.csv')

```

```python
product_dataset.head()
# product_dataset.shape
# product_dataset.info()

```

```python
new_product_dataset=product_dataset[['Id','CategoryName','ProductName','Description']]
new_product_dataset.head()


```

```python
new_product_dataset.dropna(inplace=True)
new_product_dataset.isnull().sum()
```

```python
new_product_dataset.duplicated().sum()
```

To remove special characters

```python
new_product_dataset['Description'] = new_product_dataset['Description'].str.replace(r'[^\w\s]', '', regex=True)
new_product_dataset['CategoryName'] = new_product_dataset['CategoryName'].str.replace(r'[^\w\s]', '', regex=True)
new_product_dataset['ProductName'] = new_product_dataset['ProductName'].str.replace(r'[^\w\s]', '', regex=True)
```

```python
new_product_dataset['Description'].head().tolist()
```

Removed spaces from category name and product name

```python
new_product_dataset['CategoryName'] = new_product_dataset['CategoryName'].str.replace(' ', '', regex=False)
new_product_dataset['ProductName'] = new_product_dataset['ProductName'].str.replace(' ', '', regex=False)
```

Combined category name, product name and description to make tags

```python
new_product_dataset['tags'] = new_product_dataset['CategoryName'] + " " + new_product_dataset['ProductName'] + " " + new_product_dataset['Description']
# new_product_dataset.head()
```

Make new dataset with necessary cleaned columns

```python
final_dataset = new_product_dataset[['Id', 'tags']].copy()
final_dataset.head()
```

Changed tags letters to lowercase

```python
final_dataset['tags'] = final_dataset['tags'].apply(lambda x: x.lower())
final_dataset.head()
```

Reduced words to their root base form (e.g., 'connecting' and 'connected' both become 'connect')

```python
import nltk

from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)

final_dataset['tags']=final_dataset['tags'].apply(stem)
final_dataset.head()
```

We ignore filler words (like 'the' or 'and') and only keep the top 5,000 most important words.

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')

vectors=cv.fit_transform(final_dataset['tags']).toarray()
print(vectors.shape)
```

Compare every product against every other product to calculate a "match score" (from 0 to 1) based on how many words they share in common.

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
print(similarity.shape)
```

Loop through every product, find its top 5 closest matches based on the similarity scores,
and save the results into a lightweight JSON file for the backend to use.

```python
import json

recommendations_dict = {}

for index in range(final_dataset.shape[0]):

    current_product_id = str(final_dataset.iloc[index]['Id'])

    similarity_scores = similarity[index]

    top_5_matches = sorted(list(enumerate(similarity_scores)), reverse=True, key=lambda x: x[1])[1:6]

    recommended_ids = []
    for match in top_5_matches:
        matched_index = match[0]
        matched_id = int(final_dataset.iloc[matched_index]['Id'])
        recommended_ids.append(matched_id)

    recommendations_dict[current_product_id] = recommended_ids

with open('precomputed_recommendations.json', 'w') as json_file:
    json.dump(recommendations_dict, json_file, indent=4)

print("precomputed_recommendations.json created successfully!")
```

Loaded the generated JSON file and run a quick test.
This simulates exactly what our backend will do: look up an ID, instantly get the 5 matching IDs, and fetch their names to display.

```python
import json

with open('precomputed_recommendations.json', 'r') as file:
    loaded_recommendations = json.load(file)

def test_recommendation_with_names(product_id):
    product_id_str = str(product_id)

    if product_id_str in loaded_recommendations:

        original_product = new_product_dataset[new_product_dataset['Id'] == product_id]
        original_name = original_product['ProductName'].values[0]

        print(f"Success! Top 5 recommendations for '{original_name}' (ID: {product_id}):\n")

        recommended_ids = loaded_recommendations[product_id_str]

        for rank, rec_id in enumerate(recommended_ids, start=1):

            matched_row = new_product_dataset[new_product_dataset['Id'] == rec_id]

            matched_name = matched_row['ProductName'].values[0]

            print(f"{rank}. ID: {rec_id} | Name: {matched_name}")

    else:
        print(f"Error: Product ID '{product_id}' not found in the JSON file.")

test_id = 235
test_recommendation_with_names(test_id)
```
